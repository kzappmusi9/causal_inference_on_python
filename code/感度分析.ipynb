```python
import pandas as pd
import numpy as np
import sensemakr as smkr
import statsmodels.formula.api as smf
from statsmodels.iolib.summary2 import summary_col

#サンプルデータの作成
np.random.seed(1)
n = 1000
DigitalAdSpend = np.random.normal(20, 5, n)
CustomerAge = np.random.randint(20, 50, n)
PrevPurchase = np.random.choice([0, 1], n)
Region = np.random.choice([1,2,3], n)

#サンプル売り上げデータの作成
SalesRevenue = 100 + 3 * DigitalAdSpend + 0.5 * CustomerAge + 2 * PrevPurchase + np.random.normal(0, 10, n)

df = pd.DataFrame({
    'SalesRevenue': SalesRevenue,
    'DigitalAdSpend': DigitalAdSpend,
    'CustomerAge': CustomerAge,
    'PrevPurchase': PrevPurchase,
    'Region': Region
})

#モデリング
reg_model = smf.ols('SalesRevenue ~ DigitalAdSpend + CustomerAge + PrevPurchase + Region', data=df)
sales_model = reg_model.fit()

# 感度分析の実行

"""解析結果の解釈
    - RV: Robustness Value
    - ここでは未観測の交絡因子が、処置効果の推定値をゼロ（パラメータ`q`=1の場合）にするために必要な強さを示す指標
    - 例えば、RV=0.5の場合、未観測の交絡因子が処置効果の推定値をゼロにするためには、処置と結果の両方に対して0.5以上の強さを持つ必要があることを意味する
    - `benchmark_covariates`として`CustomerAge`を指定しているため、RVの解釈は「未観測の交絡因子が、処置効果の推定値をゼロにするためには、CustomerAgeと同等以上の強さを持つ必要がある」ということになる
"""

sales_sense = smkr.Sensemakr(
    
    
    
    model=sales_model,
    treatment='DigitalAdSpend',
    benchmark_covariates=['CustomerAge'],
    kd = [1, 2, 3],
    ky = [1, 2, 3],
    q = 1,
    alpha = 0.05,
    reduce = True)

#sales_sense.summary()
# Gets HTML code and table
result=sales_sense.ovb_minimal_reporting() 
# Prints raw html code
result
```
